```python
# adding axis labels and titles

np.random.seed(613)
x = np.arange(50)
y = np.random.randint(0, 100, 50)

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)
ax.set_title('Total growth over time')
ax.set_ylabel('Total Growth')
ax.set_xlabel('Years since start')
```

```python
#Modifying labels and titles

#made a dictionary with the specifications of each font
font1 = {'family':'sans-serif','color':'black','size':20}
font2 = {'family':'monospace','color':"#9F6AE5",'size':14}

#add the font to the AXES
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y)

ax.set_title ('Total growth over time', fontdict = font1)
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since Start', font2)
```

```python
#To change the location of the title, in this example it is to the left.

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y)
ax.set_title('Total growth over time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)
```

```python
# To modify data points in color and shape: we can search for specific comand on google. in this example * is a star.
# We can also search for the colors on internet. 
# colors could be by the name or by the number
# we can change the style of the line and the width size. 
# we can add grid lines: it could be vertical or horizontal lines. 

fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y, 
           marker = '*', 
           markersize = 12,
           color = 'cyan',
           linestyle = '--',
           linewidth = 2,
           markeredgecolor = '#bcbd22',
           markerfacecolor = '#7f7f7f')

# labels
ax.set_title('Total growth over time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total growth', fontdict = font2)
ax.set_xlabel('Years since start', fontdict = font2)

ax.grid(axis = 'y')
```

```python
fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y, 
           marker = '*', 
           markersize = 12,
           color = "#7AD1E3",
           linestyle = '--',
           linewidth = 2,
           markeredgecolor = "#131416",
           markerfacecolor = "#ede973")

ax.grid(axis = 'y', color = "#C3D1EF", linewidth = 2, linestyle = '-.')
```

```python
# Marker is the shape of the point and you can change the size with markersize
#markeredgecolor change the border of the marker
#markerfacecolor change the inside of the marker
# Colors with # has more options. Also, in MAC you could select the spectrum in an interactive mode. 
# you could search for different shapes of markers on google
# in this AXES the labels font directory it is on a previous line.
# you could select different kind of lines: --, -., etc. search on google.



fig, ax = plt.subplots(figsize=(5,3))
ax.plot(x,y, 
           marker = 'p', 
           markersize = 7,
           color = '#B0C4DE',
           linestyle = '-.',
           linewidth = 2,
           markeredgecolor = "#17a9f2",
           markerfacecolor = "#bde6e9")

# labels
ax.set_title('Total Growth Over Time', fontdict = font1, loc = 'left')
ax.set_ylabel('Total Growth', fontdict = font2)
ax.set_xlabel('Years Since Start', fontdict = font2)

ax.grid(axis = 'y')
```
