```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0, 100,50)
y2 = np.random.randint(0, 100,50)
```

# adding and modify legend
Add a ‘label’ argument to each plot object
Call ‘legend’ and specify our legend location
By adding arguments to our legend object, we can change different elements of our legend

```python

# adding legend

fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1" )
ax.plot(x,y2, label = "Person 2" )
ax.legend(loc= "lower right",
          frameon = True,
          fontsize =12,
          ncol = 2,
          shadow = True)
fig.show()

#Call ‘legend’ and specify our legend location "lower right"
```

Moving the legend outside of the plot area
We can position the legend outside of our plot by
Specifying legend location as before
Specifying where we want to anchor that location on our figure using
bbox_to_anchor

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.plot(x,y1, label = "Person 1")
ax.plot(x,y2, label = "Person 2")
ax.legend(loc='upper right',
          bbox_to_anchor = (1,1))

#with upper right the legend is inside the frame, 
# with upper left, the legend box is outside
# bbox_to_anchor to control the location of the legend

```

```python
fig, ax = plt.subplots(figsize=(5, 3))
ax.scatter(x,y1, label = "Person 1")
ax.scatter(x,y2, label = "Person 2")
ax.legend(loc='lower right')

ax.text(10, 95, "This value is important!",
        ha = 'center',
        color = 'red',
        size = 20)
fig.show()
```

Positioning text annotations
• We can do this using transforms
• Start by making a simple, empty set of axes

```python
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])

ax.text(1,5, ".Data: (1, 5)", transform=ax.transData)
ax.text(0.5,0.1, ".Axes: (0.5, 0.1)",transform=ax.transAxes)
ax.text(0.2,0.2, ".Figure:(0.2, 0.2)", transform=fig.transFigure)

#Data, Axes and Figure is the point of reference, and the numbers inside the () is de cohordenadas from that point of reference.
# Data is the numbers of the location in the cartesian plane: X1 and Y5 for example
#Figure the canvas 
#AXES the frame of the graphic, it is inside of the figure
```

Activity: customizing our plots

```python
np.random.seed(613)
x1 = np.arange(50)
y1 = np.random.randint(0, 75,50)
x2 = np.array(["Luffy", "Zoro", "Nami", "Usopp", "Sanji"])
y2 = np.array([110, 180, 240, 99, 220])

plt.style.use('default')

fig, (ax1, ax2) = plt.subplots(ncols=2,
                               nrows=1,
                               figsize=(7,3),
                               layout='constrained')


ax1.scatter(x1,y1,
            color='#04D8B2',
            marker='p')

ax2.bar(x2,y2,
        color = '#7FFFD4')

ax1.text(10, 80, 'Graphic 1',
        ha = 'center',
        color = 'indigo',
        size = 15,
        )

ax1.set_ylabel('#Patients', color = '#DDA0DD')
ax1.set_xlabel('Time/Days', color = '#DDA0DD')

ax1.annotate('Patient 0', xy = (0,10), xytext =(20, 60),
            arrowprops=dict(facecolor='cyan',
                            arrowstyle = 'wedge'))

                        


ax2.text(0.5, 270, 'Graphic 2',
        ha = 'center',
        color = 'indigo',
        size = 15)

ax2.set_ylabel('#Patients', color = '#DDA0DD')
ax2.set_xlabel('Cluster', color = '#DDA0DD')


ax2.annotate('TOP', xy = (2.3,240), xytext =(3, 235),
            arrowprops=dict(facecolor='cyan',
                            arrowstyle = 'wedge'))
plt.xticks(rotation=45, ha='right')


```

# multiple subplots in the figure (canvas)

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize=(7,4))
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)
```

Layouts: 
We can use layouts to make sure that our subplots fit neatly in our figure area.
There are two main kinds of layout we consider:
- Tight layout adjusts subplots so tick labels, axis labels, and titles don't overlap or leave the figure area
- Constrained layout works similarly except it also fits things like legends or colorbars

```python
fig, someaxes = plt.subplot_mosaic([['ax1', 'ax3'],
                                    ['ax2', 'ax3']],
                                    figsize=(7,4),
                                    layout='constrained')
someaxes['ax1'].scatter(x1,y1)
someaxes['ax2'].bar(x2,y2)
someaxes['ax3'].plot(x1,y1)

someaxes ['ax1'].set_xlabel('A Big Label', fontsize=18)
someaxes['ax2'].set_xlabel('Another Label', fontsize=18)
someaxes['ax3'].set_xlabel('Label 2: 2 Fast 2 Furious', fontsize=18)
```

Multiple viz on one axes object

```python
x = np.array(['Luffy', 'Zoro', 'Freddie', 'Ussop', 'Sanji' ])
y1= np.array([110, 180, 240, 99, 220])
y2= np.array([170, 100, 90, 120, 50])

# define our figure and axes (just one this time)
fig, ax = plt.subplots(figsize=(7, 3))

# now call both bar and plot elements to the same axes (ax)
ax.bar(x, y1, color = 'indigo')
ax.plot(x, y2, color = 'red')
```

# Add error information
- Then use errorbar() to add in our error line (standard deviation in this case, but could be whatever value you calculated)
- yerr specifies that we’re plotting vertical error bars
- fmt makes sure we’re not plotting the actual data points, only the error

```python
#Add Error information

#First, calculate standard deviation of our data
y2_sd = np.std(y2)

#Then plot our line as before
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color = 'red')
ax.errorbar(x,y2,
            yerr=y2_sd,
            fmt ='none',
            ecolor = 'indigo',
            elinewidth = 2,
            capsize =4,
            capthick = 2 )

#capsize is the horizontal size of the hat in the vertical lines
#capthick is the vertical size of the hat in the vertical lines

```
